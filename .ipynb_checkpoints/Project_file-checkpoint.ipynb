### importing libraries

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

### Reading Dataset file 

```python
df = pd.read_csv("tcc_ceds_music.csv")
cols = [
    "artist_name",
    "track_name",
    "release_date",
    "genre",
    "danceability",
    "energy",
    "valence",
    "sadness",
    "acousticness",
    "loudness",
    "instrumentalness",
    "topic",
    "romantic",
    "obscene",
    "lyrics"
]

music_df = df[cols].copy()

music_df.head(5)
```

### Our Approach 
#### Our plan is to use the idea of divide and conquer. We are going to divide the datasets into 3 groups depending on the release date. Group 1 (1950-1980), Group 2 (1980 - 2000) and Group 3 (2000-2019) 

```python
query1 = 'release_date >= 1950 and release_date < 1980'
query2 = 'release_date >=1980 and release_date < 2000'
query3 = 'release_date >=2000 and release_date <= 2019'

group1_df = music_df.query(query1).copy()
group2_df = music_df.query(query2).copy()
group3_df = music_df.query(query3).copy()
```

### Q1. Has music become happier or sadder?

Instead of immediately grouping songs into decades, let's calculate averages for every year.

```python
yearly_emotion = group2_df.groupby("release_date")[["valence","sadness"]].mean()
yearly_emotion.reset_index()
yearly_emotion.head(10)
```

#### Line plot: visualizing Trends in Musical Happiness (1980–1999)

```python
fig,ax = plt.subplots(1,1, figsize = (10,5))

ax.plot(yearly_emotion.index, yearly_emotion.valence, color="royalblue")

ax.set_title("Evolution of Musical Happiness (1980–1999)")
ax.set_xlabel("Release Year")
ax.set_ylabel("Average Valence")
plt.xticks(range(1980, 2000, 5))
plt.grid(alpha=0.3)
plt.show()
```

#### Line plot: visualizing Trends in Musical Sadness (1980–1999)

```python
fig,ax = plt.subplots(1,1, figsize = (10,5))

ax.plot(yearly_emotion.index, yearly_emotion.sadness, color="purple")

ax.set_title("Evolution of Musical Sadness (1980–1999)")
ax.set_xlabel("Release Year")
ax.set_ylabel("Average Sadness Score")
plt.xticks(range(1980, 2000, 5))
plt.grid(alpha=0.3)
plt.show()
```

#### Bar chart: compare the 1980s with the 1990s

```python
group2_df = group2_df.copy()

group2_df["decade"] = (
    group2_df["release_date"] // 10
) * 10
```

```python
decade_emotion = group2_df.groupby("decade")[
    ["valence", "sadness"]
].mean()

print(decade_emotion.round(3))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

sns.barplot(
    data=group2_df,
    x="decade",
    y="valence",
    ax=axes[0],
    color="royalblue"
)

axes[0].set_title("Average Musical Valence")
axes[0].set_xlabel("Decade")
axes[0].set_ylabel("Average Valence")
axes[0].set_ylim(0, 1)

sns.barplot(
    data=group2_df,
    x="decade",
    y="sadness",
    ax=axes[1],
    color="purple"
)

axes[1].set_title("Average Musical Sadness")
axes[1].set_xlabel("Decade")
axes[1].set_ylabel("Average Sadness")
axes[1].set_ylim(0, 1)

plt.tight_layout()
plt.show()
```

```python
change = decade_emotion.loc[1990] - decade_emotion.loc[1980]

print("Change in Valence:", round(change["valence"], 3))
print("Change in Sadness:", round(change["sadness"], 3))
```

#### Violin Plot: How Did the Distribution of Musical Happiness Change?

```python
# Divide 1980–1999 into four 5-year periods
group2_df["era"] = pd.cut(
    group2_df["release_date"],
    bins=[1979, 1984, 1989, 1994, 1999],
    labels=["1980–84", "1985–89", "1990–94", "1995–99"]
)

# Create violin plot
plt.figure(figsize=(10, 5))

sns.violinplot(
    data=group2_df,
    x="era",
    y="valence",
    hue="era",
    palette="YlOrBr",
    inner="quartile",
    legend=False,
    cut=0
)

plt.title("Distribution of Musical Happiness (1980–1999)")
plt.xlabel("Five-Year Period")
plt.ylabel("Valence Score")
# plt.ylim(0, 1)

plt.tight_layout()
plt.show()
```

```python
summary = group2_df.groupby("era", observed=True)["valence"].agg(
    count="count",
    q1=lambda x: x.quantile(0.25),
    median="median",
    q3=lambda x: x.quantile(0.75),
    mean="mean"
)

print(summary.round(3))
```

#### Slope Chart 

```python

# Separate the 1980s and 1990s
early = group2_df[
    group2_df["release_date"].between(1980, 1989)
]

late = group2_df[
    group2_df["release_date"].between(1990, 1999)
]

# Calculate average valence by genre
early_avg = early.groupby("genre")["valence"].mean()
late_avg = late.groupby("genre")["valence"].mean()

# Count songs in each genre
early_count = early.groupby("genre")["valence"].count()
late_count = late.groupby("genre")["valence"].count()

# Combine results
comparison = pd.DataFrame({
    "1980s": early_avg,
    "1990s": late_avg,
    "count_1980s": early_count,
    "count_1990s": late_count
})

# Keep genres with at least 10 valid songs in each decade
comparison = comparison[
    (comparison["count_1980s"] >= 10) &
    (comparison["count_1990s"] >= 10)
].copy()

# Calculate the change in valence
comparison["change"] = (
    comparison["1990s"] - comparison["1980s"]
)

# Sort by valence change
comparison = comparison.sort_values("change")

# Create slope chart
fig, ax = plt.subplots(figsize=(10, 7))

for genre, row in comparison.iterrows():

    # Green = happier, blue = sadder
    color = "seagreen" if row["change"] > 0 else "steelblue"

    ax.plot(
        [0, 1],
        [row["1980s"], row["1990s"]],
        marker="o",
        color=color,
        linewidth=2.5
    )

    # Genre and average valence in the 1980s
    ax.text(
        -0.05,
        row["1980s"],
        f'{genre}: {row["1980s"]:.2f}',
        ha="right",
        va="center",
        fontsize=10
    )

    # Average valence and change in the 1990s
    ax.text(
        1.05,
        row["1990s"],
        f'{row["1990s"]:.2f} ({row["change"]:+.2f})',
        ha="left",
        va="center",
        fontsize=10
    )

ax.set_xticks([0, 1])
ax.set_xticklabels(["1980s", "1990s"], fontsize=12)

ax.set_xlim(-0.5, 1.5)
ax.set_ylim(0, 1)

ax.set_ylabel("Average Valence")

ax.set_title(
    "How Did Musical Happiness Change by Genre?\n1980s vs. 1990s",
    fontsize=14,
    fontweight="bold"
)

ax.grid(axis="y", alpha=0.2)

for spine in ["top", "right", "left"]:
    ax.spines[spine].set_visible(False)

plt.tight_layout()
plt.show()
```

#### General Summary: Has Music Become Happier or Sadder? (1980–1999)

Our analysis suggests that music became less positive-sounding between 1980 and 1999, although the changes were gradual rather than dramatic.

The line plot allows us to examine how average musical valence changed year by year. Together with the violin plot, it helps us investigate the overall direction of musical happiness across the two decades. However, the exact yearly trend should be confirmed from the line plot's calculated values.

The violin plot provides clearer evidence of a downward shift in musical positivity. The median valence decreased from approximately 0.62 in 1980–1984 to 0.52 in 1995–1999. The distribution also shifted toward lower valence scores, suggesting that less positive-sounding songs became more prominent. Nevertheless, highly positive-sounding songs remained present throughout both decades.

The slope chart helps us investigate whether this change occurred across all music genres or was concentrated in particular genres. By comparing average valence in the 1980s and 1990s, it shows which genres became more or less positive-sounding. The actual slope chart results are needed to identify the genres responsible for the largest changes.

##### Overall Conclusion

Our findings indicate a gradual decline in musical positivity from the 1980s to the 1990s, with the distribution of valence shifting toward lower scores. However, this does not necessarily mean that songs became lyrically sadder, since valence measures musical mood rather than sadness in lyrics. The genre-level comparison can help determine whether the overall decline was widespread or driven by particular musical styles.
