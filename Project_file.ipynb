### importing libraries

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

### Reading Dataset file 

```python
df = pd.read_csv("tcc_ceds_music.csv")
cols = [
    "artist_name",
    "track_name",
    "release_date",
    "genre",
    "danceability",
    "energy",
    "valence",
    "sadness",
    "acousticness",
    "loudness",
    "instrumentalness",
    "topic",
    "romantic",
    "obscene",
    "lyrics"
]

music_df = df[cols].copy()

music_df.head(5)
```

### Our Approach 
#### Our plan is to use the idea of divide and conquer. We are going to divide the datasets into 3 groups depending on the release date. Group 1 (1950-1980), Group 2 (1980 - 2000) and Group 3 (2000-2019) 

```python
query1 = 'release_date >= 1950 and release_date < 1980'
query2 = 'release_date >=1980 and release_date < 2000'
query3 = 'release_date >=2000 and release_date <= 2019'

group1_df = music_df.query(query1).copy()
group2_df = music_df.query(query2).copy()
group3_df = music_df.query(query3).copy()
```

**Project Theme:** How Have the Sound, Mood, and Lyrics of Music Changed Over 70 Years?

## Project Overview

This project explores how music has evolved from 1950 to 2019 by examining changes in musical emotions, energy, danceability, lyrical themes, and sound characteristics.
